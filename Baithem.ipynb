```python
from baithem import *

def main_menu():
    danhSach = docTaiKhoanTuCSV()

    while True:
        print("\n" + "="*40)
        print("HỆ THỐNG QUẢN LÝ NGÂN HÀNG (Console)")
        print("="*40)
        print("1. Tạo tài khoản mới")
        print("2. Gửi tiền")
        print("3. Rút tiền")
        print("4. Xem tất cả tài khoản")
        print("5. Sao lưu dữ liệu")
        print("6. Thoát")
        print("="*40)

        choice = input("Chọn chức năng (1-6): ").strip()

        if choice == "1":
            tao_tai_khoan(danhSach)
        elif choice == "2":
            gui_tien(danhSach)
        elif choice == "3":
            rut_tien(danhSach)
        elif choice == "4":
            xem_tat_ca(danhSach)
        elif choice == "5":
            sao_luu_du_lieu()
        elif choice == "6":
            print("👋 Tạm biệt!")
            break
        else:
            print("❌ Lựa chọn không hợp lệ!")

if __name__ == "__main__":
    main_menu()
```
