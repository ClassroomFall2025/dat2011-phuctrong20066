## Lab 5 - Bài 1

```python
# Code lab 5 bài 1 ở đây
class Sanpham:
    
    def __init__(self, ten_san_pham, gia, giam_gia):
        self.ten_san_pham = ten_san_pham
        self.gia = gia
        self.giam_gia = giam_gia
    def nhap_thong_tin(self):
        self.ten_san_pham = input("Nhập tên sản phẩm: ")
        self.gia = float(input("Nhập giá sản phẩm: "))
        self.giam_gia= float(input("Giảm giá: "))
    def thue_nhap_khau(self):
        return self.gia * 0.1
    def xuat_thong_tin_sp(self):
        print(f"Tên sản phẩm: {self.ten_san_pham}")
        print(f"Giá: {self.gia}")
        print(f"Giảm giá: {self.giam_gia}")    
        print(f"Thuế nhập khẩu: {self.thue_nhap_khau()}")
        
sp1 = Sanpham("Điện thoại", 2000000, 1000000)
print(sp1)
sp1.xuat_thong_tin_sp()
```
