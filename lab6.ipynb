## Lab 6 - Bài 4

```python
# code lab 6 bài 4 ở đây
from lab62 import SinhVienPoly

ds_sv = []

while True:
    print("""
=== QUẢN LÝ SINH VIÊN ===
1. Nhập DS sinh viên
2. Xuất DS sinh viên
3. DS sinh viên Giỏi
4. Sắp xếp theo điểm giảm dần
5. Thoát
""")
    chon = input("→ Chọn (1-5): ")

    if chon == "1":
        for i in range(int(input("Nhập số lượng SV: "))):
            print(f"\n--- SV {i+1} ---")
            ho_ten, nganh = input("Họ tên: "), input("Ngành: ")
            diem = float(input("Điểm: "))
            sv = SinhVienPoly(ho_ten, nganh)
            sv.get_Diem = lambda diem=diem: diem
            ds_sv.append(sv)

    elif chon == "2":
        print("\n=== DANH SÁCH SINH VIÊN ===")
        [sv.xuat() for sv in ds_sv]

    elif chon == "3":
        print("\n=== SINH VIÊN GIỎI ===")
        [sv.xuat() for sv in ds_sv if sv.get_HocLuc() == "Giỏi"]

    elif chon == "4":
        print("\n=== SẮP XẾP THEO ĐIỂM GIẢM DẦN ===")
        for sv in sorted(ds_sv, key=lambda sv: sv.get_Diem(), reverse=True): sv.xuat()

    elif chon == "5":
        print("Tạm biệt!"); break
    else:
        print(" Lựa chọn không hợp lệ.")
```
