## Lab 6 - Bài 4

```python
# Code lab 6  bài 4 ở đây
from lab62 import SinhVienPoly

ds_sv = []

while True:
    print("""
=== QUẢN LÝ SINH VIÊN ===
1. Nhập DS sinh viên
2. Xuất DS sinh viên
3. DS sinh viên Giỏi
4. Sắp xếp theo điểm giảm dần
5. Thoát
""")
    chon = input("→ Chọn (1-5): ")

    if chon == "1":
        n = int(input("Nhập số lượng SV: "))
        for i in range(n):
            print(f"\n--- Sinh viên {i+1} ---")
            ho_ten = input("Họ tên: ")
            nganh = input("Ngành: ")
            diem = float(input("Điểm: "))

            sv = SinhVienPoly(ho_ten, nganh)
            sv.get_Diem = lambda diem=diem: diem
            ds_sv.append(sv)

        print(f" Đã nhập xong {n} sinh viên!")

    elif chon == "2":
        if not ds_sv:
            print(" Danh sách rỗng!")
        else:
            print("\n=== DANH SÁCH SINH VIÊN ===")
            for sv in ds_sv:
                sv.xuat()

    elif chon == "3":
        print("\n=== SINH VIÊN GIỎI ===")
        ds_gioi = [sv for sv in ds_sv if sv.get_HocLuc() == "Giỏi"]
        if ds_gioi:
            for sv in ds_gioi:
                sv.xuat()
        else:
            print(" Không có sinh viên giỏi nào.")

    elif chon == "4":
        print("\n=== SẮP XẾP THEO ĐIỂM GIẢM DẦN ===")
        if ds_sv:
            ds_sv_sorted = sorted(ds_sv, key=lambda sv: sv.get_Diem(), reverse=True)
            for sv in ds_sv_sorted:
                sv.xuat()
        else:
            print(" Danh sách rỗng!")

    elif chon == "5":
        print(" Tạm biệt!")
        break

    else:
        print("⚠️ Lựa chọn không hợp lệ. Vui lòng chọn 1-5.")
```
