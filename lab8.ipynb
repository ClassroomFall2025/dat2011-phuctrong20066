## Lab 8 - Bài 4

```python
# code lab 8 bài 4 ở đây
import statistics
try:
    f = open(r"H:\Python\so_ngau_nhien.txt.txt", "r", encoding="utf-8")
    data = f.read().split()
    f.close()

    if len(data) == 0:
        print(" File không có dữ liệu hoặc đường dẫn sai!")
    else:
        numbers = [int(x) for x in data]
        mean_value = statistics.mean(numbers)
        std_value = statistics.stdev(numbers)
        print("Tổng số phần tử:", len(numbers))
        print("Giá trị trung bình (Mean):", mean_value)
        print("Độ lệch chuẩn (Standard Deviation):", std_value)

except FileNotFoundError:
    print("Không tìm thấy file! Hãy kiểm tra lại đường dẫn H:\\Python\\Lab7-Exercise3.txt")
```

## Lab 8 - Bài 5: bài tập làm thêm

```python
# code lab 8  bài làm thêm ở đây 
# Kiểm tra mã số sinh viên hợp lệ
import re
def kiem_tra_ma_so_sinh_vien(ma_so):
    pattern = r'^[0-9]{5}$'
    if re.match(pattern, ma_so):
        return True
    else:
        return False
ma_so_sinh_vien = input("Nhập mã số sinh viên (8 chữ số): ")
if kiem_tra_ma_so_sinh_vien(ma_so_sinh_vien):
    print("Mã số sinh viên hợp lệ.")
else:
    print("Mã số sinh viên không hợp lệ. Vui lòng nhập đúng định dạng 8 chữ số.")
```
